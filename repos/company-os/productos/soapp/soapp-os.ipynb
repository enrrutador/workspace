# SO-app: Ubuntu en Kaggle (una sola celda)

Levanta el sistema operativo completo: rootfs + Chrome + escritorio visible,
lo publica por tunel y lo respalda en un Dataset.
Abre el link `.../vnc.html` en el telefono. Toque = clic.


```python
import os
import re
import shutil
import socket
import subprocess
import tarfile
import time
import urllib.request

# PATH extendido, jamas reemplazado: un PATH vacio ya nos rompio todo una vez
# y sin /opt/conda/bin no se resuelven npm/node/kaggle.
os.environ["PATH"] = ("/usr/local/sbin:/usr/local/bin:/usr/sbin:/usr/bin:/sbin:/bin:"
                      + os.environ.get("PATH", ""))
os.environ["LC_ALL"] = "C"
os.environ["LANG"] = "C"

DIR = "/kaggle/working/vmdata/proot"
ROOTFS = os.path.join(DIR, "rootfs")
PORT = 4096
PROOT = "/usr/bin/proot"
APTGET = "/usr/bin/apt-get"
DPKG = "/usr/bin/dpkg"
DPKGDEB = "/usr/bin/dpkg-deb"
LDCONFIG = "/sbin/ldconfig"
TIMEOUT = "/usr/bin/timeout"


def log(m=""):
    print(m, flush=True)


def run(cmd, cwd="/", check=False):
    log("$ " + " ".join(cmd))
    p = subprocess.Popen(cmd, cwd=cwd, stdout=subprocess.PIPE,
                         stderr=subprocess.STDOUT, text=True, errors="replace")
    for line in p.stdout:
        print(line, end="", flush=True)
    p.wait()
    if check and p.returncode != 0:
        raise RuntimeError("fallo rc=%d: %s" % (p.returncode, " ".join(cmd)))
    return p.returncode


def grab(cmd, cwd="/"):
    r = subprocess.run(cmd, cwd=cwd, stdout=subprocess.PIPE,
                       stderr=subprocess.STDOUT, text=True, errors="replace")
    return r.returncode, r.stdout


def proot(*args):
    return [PROOT, "-R", ROOTFS, "-b", "/dev", "-b", "/proc",
            "-b", "/sys", "-w", "/"] + list(args)


APT = [APTGET, "-o", "APT::Sandbox::User=root", "-o", "Acquire::ForceIPv4=true"]


def download(url, dest, label):
    if os.path.exists(dest) and os.path.getsize(dest) > 1000000:
        if dest.endswith(".deb") and not _deb_ok(dest):
            log("%s: corrupto, re-descargando..." % label)
            os.remove(dest)
        else:
            log("%s: existente, reuso (%dMB)" % (label, os.path.getsize(dest) // 1048576))
            return
    log("%s: descargando..." % label)
    req = urllib.request.Request(url, headers={"User-Agent": "soapp/0.1"})
    with urllib.request.urlopen(req, timeout=60) as r, open(dest, "wb") as f:
        total = int(r.headers.get("Content-Length", 0))
        got, last = 0, 0
        while True:
            chunk = r.read(4 * 1024 * 1024)
            if not chunk:
                break
            f.write(chunk)
            got += len(chunk)
            if got - last > 20 * 1024 * 1024:
                extra = ("/%dMB" % (total // 1048576)) if total else ""
                log("  ...%dMB%s" % (got // 1048576, extra))
                last = got
    if dest.endswith(".deb") and not _deb_ok(dest):
        raise SystemExit("ERROR: %s llego corrupto dos veces" % label)
    log("%s: OK (%dMB)" % (label, os.path.getsize(dest) // 1048576))


def _deb_ok(path):
    if not os.path.exists("/usr/bin/dpkg-deb"):
        return True
    r = subprocess.run(["/usr/bin/dpkg-deb", "--info", path],
                       stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
    return r.returncode == 0


def port_open(port):
    try:
        socket.create_connection(("127.0.0.1", port), timeout=2).close()
        return True
    except OSError:
        return False


os.makedirs(DIR, exist_ok=True)
os.makedirs("/kaggle/working", exist_ok=True)

log("=== [1/6] proot ===")
if not os.path.exists(PROOT):
    log("instalando proot...")
    run(["/usr/bin/apt-get", "update"], check=True)
    run(["/usr/bin/apt-get", "install", "-y", "proot"], check=True)
if not os.path.exists(PROOT):
    raise SystemExit("ERROR: no hay proot ni se pudo instalar")
run([PROOT, "--version"])

log("=== [2/6] rootfs ===")
MARK = os.path.join(ROOTFS, ".soapp-ok")
if os.path.isfile(MARK):
    log("rootfs existente, reuso")
elif os.path.isfile(os.path.join(ROOTFS, "bin", "sh")):
    open(MARK, "w").write("ok\n")
    log("rootfs previo sano, reuso")
else:
    download("https://cdimage.ubuntu.com/ubuntu-base/releases/22.04/release/ubuntu-base-22.04-base-amd64.tar.gz",
             os.path.join(DIR, "ubuntu-base.tar.gz"), "ubuntu-base")
    os.makedirs(ROOTFS, exist_ok=True)
    log("extrayendo...")
    with tarfile.open(os.path.join(DIR, "ubuntu-base.tar.gz")) as t:
        t.extractall(ROOTFS)
    open(MARK, "w").write("ok\n")
    log("rootfs OK")
if run(proot("/bin/sh", "-c", "command -v curl")) != 0:
    log("apt update + curl (1-2 min, tope 5 min por comando)...")
    run([TIMEOUT, "300"] + proot("/bin/sh", "-c", " ".join(APT + ["update"])))
    run([TIMEOUT, "300"] + proot("/bin/sh", "-c", " ".join(APT + ["install", "-y", "curl", "ca-certificates"])))
else:
    log("curl existente, reuso")
run(proot("/bin/sh", "-c", "command -v curl && echo CURL_OK"), check=True)

log("=== [3/6] Chrome ===")
CHROME = ("LD_LIBRARY_PATH=/usr/lib/x86_64-linux-gnu "
          "/opt/google/chrome/chrome --version --no-sandbox")
if run(proot("/bin/sh", "-c", CHROME)) == 0:
    log("Chrome existente, reuso")
else:
    download("https://dl.google.com/linux/direct/google-chrome-stable_current_amd64.deb",
             "/tmp/chrome.deb", "Chrome")
    log("desempaquetando Chrome (postinst se ignora a proposito)...")
    run(proot("/bin/sh", "-c", "%s -i --force-all /tmp/chrome.deb 2>&1 | tail -1 || true" % DPKG))
    log("resolviendo dependencias con apt-cache (solo lectura)...")
    rc, out = grab(proot("/bin/sh", "-c",
                          "apt-cache depends --recurse --no-recommends --no-suggests "
                          "--no-conflicts --no-breaks --no-replaces --no-enhances "
                          "google-chrome-stable 2>&1"))
    if rc != 0 or "Depends:" not in out:
        raise SystemExit("ERROR: apt-cache no conoce google-chrome-stable:\n" + out[-2000:])
    deps = sorted({m.group(1) for m in re.finditer(r"^\s*\|?Depends:\s*(\S+)", out, re.M)
                   if not m.group(1).startswith("<")})
    log("deps: %d" % len(deps))
    run(proot("/bin/sh", "-c", "mkdir -p /tmp/debs"))
    rc, _ = grab(proot("/bin/sh", "-c", "ls /var/lib/apt/lists/*_Packages 2>/dev/null | head -1"))
    if rc != 0:
        log("listas apt ausentes, actualizando...")
        run([TIMEOUT, "300"] + proot("/bin/sh", "-c", " ".join(APT + ["update", "-qq"])), check=True)
    os.makedirs("/tmp/debs", exist_ok=True)
    missing = [p for p in deps
               if not [f for f in os.listdir("/tmp/debs")
                       if f.startswith(p + "_") and f.endswith(".deb")]]
    if missing:
        log("faltan %d, bajada en bloque (una sola invocacion)..." % len(missing))
        run([TIMEOUT, "600"] + proot(
            "/bin/sh", "-c", "cd /tmp/debs && " + " ".join(APT + ["download"] + missing)))
    still = [p for p in deps
             if not [f for f in os.listdir("/tmp/debs")
                     if f.startswith(p + "_") and f.endswith(".deb")]]
    for pkg in still:
        rc, out = grab([TIMEOUT, "180"] + proot(
            "/bin/sh", "-c", "cd /tmp/debs && " + " ".join(APT + ["download", pkg])))
        if rc != 0:
            log("NO-DL: %s" % pkg)
    log("extrayendo .deb uno por uno (errores visibles)...")
    failures = []
    for deb in sorted(os.listdir("/tmp/debs")):
        if not deb.endswith(".deb"):
            continue
        rc, out = grab(proot("/bin/sh", "-c", "%s -x /tmp/debs/%s / 2>&1" % (DPKGDEB, deb)))
        if rc != 0:
            failures.append(deb)
            log("NO-X: %s" % deb)
    run(proot("/bin/sh", "-c", "%s; echo LDCONFIG_OK" % LDCONFIG))
    if failures:
        log("extraccion fallo en %d paquetes (sigo igual)" % len(failures))
    if run(proot("/bin/sh", "-c", CHROME)) == 0:
        log("CHROME_OK")
    else:
        rc, out = grab(proot("/bin/sh", "-c",
                              "ldd /opt/google/chrome/chrome 2>/dev/null | grep 'not found' || true"))
        log("CHROME_FALLA (sigo a opencode igual), falta:\n" + out)


VNC_PORT, NOVNC_PORT = 5900, 6080

GXE = {"DISPLAY": ":0", "HOME": "/root", "LC_ALL": "C",
       "PATH": "/usr/local/sbin:/usr/local/bin:/usr/sbin:/usr/bin:/sbin:/bin"}
log("=== [4/6] escritorio Ubuntu ===")
run([TIMEOUT, "300"] + proot("/bin/sh", "-c", "apt-get -o APT::Sandbox::User=root -o Acquire::ForceIPv4=true update -qq"))
run([TIMEOUT, "600"] + proot("/bin/sh", "-c", "apt-get -o APT::Sandbox::User=root -o Acquire::ForceIPv4=true install -y tigervnc-standalone-server openbox xterm || true"))
for b in ["Xvnc", "openbox-session"]:
    rc, _ = grab(proot("/bin/sh", "-c", "command -v " + b))
    if rc != 0:
        raise SystemExit("ERROR: no quedo instalado " + b)
log("Xvnc + openbox OK")

NOVNC = os.path.join(DIR, "novnc")
if not os.path.isfile(os.path.join(NOVNC, "vnc.html")):
    log("descargando noVNC...")
    urllib.request.urlretrieve("https://github.com/novnc/noVNC/archive/refs/heads/master.tar.gz", "/tmp/novnc.tgz")
    shutil.rmtree(NOVNC, ignore_errors=True)
    os.makedirs(DIR, exist_ok=True)
    with tarfile.open("/tmp/novnc.tgz") as t:
        t.extractall("/tmp/novnc-x")
    import glob as _g
    shutil.move(_g.glob("/tmp/novnc-x/*/")[0], NOVNC)
    log("noVNC OK")
else:
    log("noVNC existente, reuso")
if shutil.which("websockify") is None:
    log("instalando websockify...")
    run(["pip", "install", "-q", "websockify"], check=True)

run(["/usr/bin/pkill", "-f", "Xvnc :0"])
run(["/usr/bin/pkill", "-f", "websockify.*6080"])
time.sleep(2)
xv = open("/tmp/xvnc.log", "w")
subprocess.Popen(proot("/bin/sh", "-c",
    "rm -f /tmp/.X0-lock; Xvnc :0 -geometry 1280x800 -depth 24 -SecurityTypes None"),
    stdout=xv, stderr=subprocess.STDOUT, start_new_session=True,
    env={**os.environ, **GXE})
ok = False
for _ in range(30):
    if port_open(VNC_PORT):
        ok = True
        break
    time.sleep(2)
if not ok:
    run(["/usr/bin/tail", "-15", "/tmp/xvnc.log"])
    raise SystemExit("ERROR: Xvnc no levanto")
log("Xvnc UP")
subprocess.Popen(proot("/bin/sh", "-c", "DISPLAY=:0 openbox-session"),
                 stdout=open("/tmp/openbox.log", "w"), stderr=subprocess.STDOUT,
                 start_new_session=True, env={**os.environ, **GXE})
time.sleep(3)
subprocess.Popen(proot("/bin/sh", "-c",
    "DISPLAY=:0 LD_LIBRARY_PATH=/usr/lib/x86_64-linux-gnu "
    "/opt/google/chrome/chrome --no-sandbox --disable-dev-shm-usage --disable-gpu "
    "--no-first-run --window-size=1280,800 about:blank"),
    stdout=open("/tmp/chrome-gui.log", "w"), stderr=subprocess.STDOUT,
    start_new_session=True, env={**os.environ, **GXE})
log("openbox + Chrome lanzados")
ws = open("/tmp/websockify.log", "w")
subprocess.Popen(["websockify", "--web", NOVNC, str(NOVNC_PORT), "127.0.0.1:%d" % VNC_PORT],
                 stdout=ws, stderr=subprocess.STDOUT, start_new_session=True)
time.sleep(3)
if port_open(NOVNC_PORT):
    log("noVNC UP")
else:
    run(["/usr/bin/tail", "-15", "/tmp/websockify.log"])
    raise SystemExit("ERROR: websockify no levanto")


log("=== [5/6] tunel al escritorio ===")
CFD = "/tmp/cloudflared"
if not os.access(CFD, os.X_OK):
    urllib.request.urlretrieve("https://github.com/cloudflare/cloudflared/releases/latest/download/cloudflared-linux-amd64", CFD)
    os.chmod(CFD, 0o755)
run(["/usr/bin/pkill", "-f", "cloudflared tunnel --url http://127.0.0.1:%d" % NOVNC_PORT])
time.sleep(2)
tlog = open("/tmp/tunnel-vnc.log", "w")
subprocess.Popen([CFD, "tunnel", "--url", "http://127.0.0.1:%d" % NOVNC_PORT],
                 stdout=tlog, stderr=subprocess.STDOUT, start_new_session=True)
URL = ""
for _ in range(45):
    with open("/tmp/tunnel-vnc.log", errors="replace") as f:
        m = re.search(r"https://[a-zA-Z0-9.-]+\.trycloudflare\.com", f.read())
    if m:
        URL = m.group(0)
        break
    time.sleep(2)
if not URL:
    run(["/usr/bin/tail", "-15", "/tmp/tunnel-vnc.log"])
    raise SystemExit("ERROR: sin URL de tunel")
log("ESCRITORIO: " + URL + "/vnc.html")
log("Abre ese link en el telefono: veras el escritorio Ubuntu con Chrome. Toque = clic.")

log("=== [6/6] respaldo a Dataset (no frena nada si falla) ===")
USER = os.environ.get("KAGGLE_USERNAME", "")
KEY = os.environ.get("KAGGLE_KEY", "")
if not USER or not KEY:
    log("BACKUP_SALTADO: faltan secrets KAGGLE_USERNAME/KAGGLE_KEY (el TUNEL de arriba sigue valiendo)")
elif shutil.which("kaggle") is None:
    log("BACKUP_SALTADO: CLI kaggle no instalado (el TUNEL de arriba sigue valiendo)")
else:
    import datetime as _dt
    import json as _json

    def _tarfilter(ti):
        return None if ti.name.endswith(".deb") else ti

    tarball = os.path.join(DIR, "soapp-rootfs.tar.gz")
    log("empaquetando rootfs (varios min sin salida, normal)...")
    with tarfile.open(tarball, "w:gz") as t:
        t.add(ROOTFS, arcname="rootfs", filter=_tarfilter)
    log("tarball: %dMB" % (os.path.getsize(tarball) // 1048576))
    dsdir = "/tmp/ds"
    shutil.rmtree(dsdir, ignore_errors=True)
    os.makedirs(dsdir, exist_ok=True)
    shutil.copy(tarball, dsdir)
    with open(os.path.join(dsdir, "dataset-metadata.json"), "w") as f:
        _json.dump({"title": "soapp-rootfs", "id": "%s/soapp-rootfs" % USER,
                    "licenses": [{"name": "CC0-1.0"}]}, f)
    rc, out = grab(["kaggle", "datasets", "create", "-p", dsdir])
    if rc != 0:
        log("create fallo, probando versionar:\n" + "\n".join(out.strip().splitlines()[-3:]))
        rc, out = grab(["kaggle", "datasets", "version", "-p", dsdir,
                        "-m", "checkpoint %s" % _dt.datetime.now(_dt.timezone.utc).strftime("%FT%TZ")])
        if rc != 0:
            log("version fallo:\n" + "\n".join(out.strip().splitlines()[-3:]))
    log("BACKUP_OK" if rc == 0 else "BACKUP_FALLO (el TUNEL de arriba sigue valiendo)")
log("ALL_DONE")
```
